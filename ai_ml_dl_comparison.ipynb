## Comparison Table
| Aspect | AI | Machine Learning | Deep Learning |
|---|---|---|---|
| Scope | Broadest field | Subset of AI | Subset of ML |
| Main idea | Intelligent task performance | Learning from data | Learning complex patterns with neural networks |
| Examples | Virtual assistants | Recommendations, spam detection | Image and speech recognition |

```python
# Veda Technology - Task 10
# Submitted by: Ayush Rajput

topics = [
    ("Artificial Intelligence", "Broad field of intelligent systems",
     "Virtual assistants, expert systems"),
    ("Machine Learning", "Subset of AI that learns patterns from data",
     "Spam detection, recommendation systems"),
    ("Deep Learning", "Subset of ML using multi-layer neural networks",
     "Image recognition, speech recognition, generative AI")
]

for name, definition, examples in topics:
    print("\n" + name)
    print("Definition:", definition)
    print("Examples:", examples)

print("\nRelationship: AI -> ML -> Deep Learning")
```

## Relationship
AI is the broader field. ML is a subset of AI, and Deep Learning is a subset of ML. Not every AI system uses ML, and not every ML system uses Deep Learning.

## Real-World Examples
- AI: virtual assistants and expert systems
- ML: spam filtering and recommendation systems
- Deep Learning: image recognition, speech recognition, generative AI

## Interview Questions
1. What is AI? — The field of building systems that perform tasks associated with human-like intelligence.
2. What is ML? — A subset of AI that learns patterns from data.
3. ML vs Deep Learning? — Deep Learning is a type of ML based on multi-layer neural networks.
